```python
import os

os.environ["PROJ_LIB"] = r"C:\Users\franz\miniconda3\envs\geoai-qa\Library\share\proj"
import rasterio
from shapely.geometry import box
import geopandas as gpd
from sqlalchemy import create_engine

tif = r"C:\temp\65a5b537c2e5aa0001f0a1f4.tif"

with rasterio.open(tif) as src:
    geom = box(*src.bounds)
    crs = src.crs

gdf = gpd.GeoDataFrame(
    [{"path": tif, "geometry": geom}],
    crs=crs
).to_crs(4326)

engine = create_engine("postgresql://postgres:2712@localhost:5432/geoai")

gdf.to_postgis("tiffs", engine, schema="src", if_exists="append", index=False)

print("TIFF registered.")
```
